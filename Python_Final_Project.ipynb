```python
# nova_university_oop.py

class Person:
    def __init__(self, name, email, phone):
        self.name = name
        self.email = email
        self.phone = phone

    def display_info(self):
        return f"Name: {self.name}, Email: {self.email}, Phone: {self.phone}"


class Student(Person):
    def __init__(self, student_id, name, email, phone, department, attendance, python_marks, sql_marks, excel_marks):
        super().__init__(name, email, phone)
        if attendance < 0 or attendance > 100:
            raise ValueError("Attendance must be between 0 and 100")
        self.student_id = student_id
        self.department = department
        self.attendance = attendance
        self.python_marks = python_marks
        self.sql_marks = sql_marks
        self.excel_marks = excel_marks

    def total_marks(self):
        return self.python_marks + self.sql_marks + self.excel_marks

    def average_marks(self):
        return self.total_marks() / 3

    def result_status(self):
        return "Pass" if self.average_marks() >= 50 else "Fail"

    def grade(self):
        avg = self.average_marks()
        if avg >= 80:
            return "A"
        elif avg >= 65:
            return "B"
        elif avg >= 50:
            return "C"
        else:
            return "F"

    def attendance_eligibility(self):
        return self.attendance >= 75

    def display_academic_info(self):
        return (f"Student ID: {self.student_id}, Dept: {self.department}, "
                f"Attendance: {self.attendance}%, Total Marks: {self.total_marks()}, "
                f"Average: {self.average_marks():.2f}, Result: {self.result_status()}, Grade: {self.grade()}")


class Faculty(Person):
    def __init__(self, faculty_id, name, email, phone, department):
        super().__init__(name, email, phone)
        self.faculty_id = faculty_id
        self.department = department

    def display_info(self):
        return f"Faculty ID: {self.faculty_id}, {super().display_info()}, Department: {self.department}"


class UniversitySystem:
    def __init__(self):
        self.students = []
        self.faculties = []

    def add_student(self, student):
        self.students.append(student)

    def view_students(self):
        for s in self.students:
            print(s.display_academic_info())

    def check_attendance(self, student_id):
        for s in self.students:
            if s.student_id == student_id:
                return f"Eligible: {s.attendance_eligibility()}"
        return "Student not found"

    def view_results(self, student_id):
        for s in self.students:
            if s.student_id == student_id:
                return s.display_academic_info()
        return "Student not found"

    def add_faculty(self, faculty):
        self.faculties.append(faculty)

    def view_faculties(self):
        for f in self.faculties:
            print(f.display_info())


def main():
    system = UniversitySystem()

    while True:
        print("\n--- Nova University Student Management ---")
        print("1. Add Student")
        print("2. View Students")
        print("3. Check Attendance Eligibility")
        print("4. View Academic Result")
        print("5. Add Faculty")
        print("6. View Faculty")
        print("7. Exit")

        choice = input("Enter choice: ")

        try:
            if choice == "1":
                sid = input("Student ID: ")
                name = input("Name: ")
                email = input("Email: ")
                phone = input("Phone: ")
                dept = input("Department: ")
                attendance = int(input("Attendance %: "))
                python_marks = int(input("Python Marks: "))
                sql_marks = int(input("SQL Marks: "))
                excel_marks = int(input("Excel Marks: "))
                student = Student(sid, name, email, phone, dept, attendance, python_marks, sql_marks, excel_marks)
                system.add_student(student)
                print("Student added successfully!")

            elif choice == "2":
                system.view_students()

            elif choice == "3":
                sid = input("Enter Student ID: ")
                print(system.check_attendance(sid))

            elif choice == "4":
                sid = input("Enter Student ID: ")
                print(system.view_results(sid))

            elif choice == "5":
                fid = input("Faculty ID: ")
                name = input("Name: ")
                email = input("Email: ")
                phone = input("Phone: ")
                dept = input("Department: ")
                faculty = Faculty(fid, name, email, phone, dept)
                system.add_faculty(faculty)
                print("Faculty added successfully!")

            elif choice == "6":
                system.view_faculties()

            elif choice == "7":
                print("Exiting...")
                break

            else:
                print("Invalid choice. Try again.")

        except ValueError as e:
            print(f"Error: {e}")
        except Exception as e:
            print(f"Unexpected error: {e}")


if __name__ == "__main__":
    main()
''
```
